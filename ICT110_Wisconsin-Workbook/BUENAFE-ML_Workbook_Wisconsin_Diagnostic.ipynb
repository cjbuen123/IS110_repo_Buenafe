### C.2 Feature Selection
**Exercise C.4 (filter).** Find features whose absolute correlation with an *earlier* feature exceeds 0.95.

```python
# TODO: upper-triangle of the absolute correlation matrix, then columns above 0.95
...

corr_matrix = X.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [column for column in upper.columns if any(upper[column] > 0.95)]

print("Features with correlation > 0.95 to an earlier feature:", to_drop)
```

**Exercise C.5 (univariate).** Rank features with `SelectKBest(f_classif)`.

```python
# TODO: fit SelectKBest(f_classif, k=10) and build a sorted Series of scores
...

selector = SelectKBest(score_func=f_classif, k=10)
selector.fit(X, y)

kbest_scores = pd.Series(selector.scores_, index=X.columns).sort_values(ascending=False)
kbest_features = list(kbest_scores.head(10).index)

print("Top 10 SelectKBest features:\n", kbest_scores.head(10))
```

**Exercise C.6 (wrapper).** Use Recursive Feature Elimination with logistic regression on standardized data to keep 10 features.

```python
# TODO: RFE(LogisticRegression, n_features_to_select=10) on standardised X; store names in rfe_features
...

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

rfe = RFE(estimator=LogisticRegression(max_iter=1000, random_state=7), n_features_to_select=10)
rfe.fit(X_scaled, y)

rfe_features = list(X.columns[rfe.support_])
print("RFE selected features:\n", rfe_features)
```

**Exercise C.7 (embedded).** Rank features by `ExtraTreesClassifier` importance and plot the top 10.

```python
# TODO: fit ExtraTrees (200 trees, random_state=7), store importances in imp, plot barh of top 10
...

et = ExtraTreesClassifier(n_estimators=200, random_state=7)
et.fit(X, y)

imp = pd.Series(et.feature_importances_, index=X.columns).sort_values(ascending=True)
extratrees_features = list(imp.tail(10).index)[::-1]

imp.tail(10).plot(kind='barh', figsize=(8, 5))
plt.title("Top 10 Features by ExtraTrees Importance")
plt.xlabel("Importance")
plt.show()
```

**Exercise C.8 (decision).** Keep a feature if **at least two of the three** methods (SelectKBest top 10, RFE, ExtraTrees top 10) chose it. Store the result in a list called `selected`.

```python
# TODO: vote across the three methods
...

from collections import Counter

all_votes = kbest_features + rfe_features + extratrees_features
vote_counts = Counter(all_votes)

selected = [feature for feature, count in vote_counts.items() if count >= 2]
print(f"Selected {len(selected)} features supported by at least 2 methods:\n", selected)
```

> **Caution (data leakage).** Here selection used the whole dataset to keep the exercise simple. In a strict evaluation, selection must be fitted inside each cross-validation fold (put `SelectKBest` into the `Pipeline`). You will discuss this in the final questions.

## D. Evaluate Algorithm

We use **stratified 10-fold cross-validation** on the training set as our test harness. Scaling is placed inside a `Pipeline` so it is re-fitted in every fold: this avoids leakage.

**Exercise D.1.** Define the six candidate models: Logistic Regression, LDA, KNN, CART, Gaussian Naive Bayes and SVM.

```python
seed = 7
kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=seed)

# TODO: dictionary named models with keys LR, LDA, KNN, CART, NB, SVM
...

models = {
    'LR': LogisticRegression(max_iter=1000, random_state=seed),
    'LDA': LinearDiscriminantAnalysis(),
    'KNN': KNeighborsClassifier(),
    'CART': DecisionTreeClassifier(random_state=seed),
    'NB': GaussianNB(),
    'SVM': SVC(random_state=seed)
}

def spot_check(models, X_tr, scale):
    """Cross-validated accuracy for every model (optionally scaled inside the pipeline)."""
    results = {}
    for name, model in models.items():
        steps = ([("scale", StandardScaler())] if scale else []) + [("model", model)]
        results[name] = cross_val_score(Pipeline(steps), X_tr, y_train,
                                        cv=kfold, scoring="accuracy")
    return results
```

**Exercise D.2.** Run the spot-check on all features, once **raw** and once **scaled**, and tabulate the mean accuracies.

```python
# TODO: call spot_check twice and build a DataFrame called summary with columns raw / scaled
...

results_raw = spot_check(models, X_train, scale=False)
results_scaled = spot_check(models, X_train, scale=True)

summary = pd.DataFrame({
    'raw': {name: scores.mean() for name, scores in results_raw.items()},
    'scaled': {name: scores.mean() for name, scores in results_scaled.items()}
})

display(summary)
```

✏️ **Question D.3.** Which models benefit from scaling, and why?

> **Your answer:**

*KNN, SVM, and Logistic Regression benefit the most from feature scaling.

Distance & Gradient-Based Models (KNN, SVM, LR): KNN relies on Euclidean distance calculations, while SVM computes margins based on feature geometry, and Logistic Regression uses gradient descent. Unscaled features with large numerical ranges (e.g., area > 1000 vs. smoothness < 0.1) dominate these distance and gradient calculations, severely distorting model training. Scaling puts all features on equal footing.

Tree-Based Models (CART): Decision trees split each feature independently based on ordinal thresholds, making them scale-invariant.*

**Exercise D.6.** In a medical setting accuracy is not enough. For the best scaled model, cross-validate **accuracy, ROC-AUC and recall for the malignant class** (`pos_label=0`).

```python
# TODO: cross_validate with a scoring dict including make_scorer(recall_score, pos_label=0)
...

scoring = {
    'accuracy': 'accuracy',
    'roc_auc': 'roc_auc',
    'recall_malignant': make_scorer(recall_score, pos_label=0)
}

best_pipeline = Pipeline([
    ('scale', StandardScaler()),
    ('model', SVC(random_state=seed))
])

cv_results = cross_validate(best_pipeline, X_train, y_train, cv=kfold, scoring=scoring)

print("Best Scaled Model (SVM) Performance:")
print(f"Mean Accuracy:         {cv_results['test_accuracy'].mean():.4f}")
print(f"Mean ROC-AUC:          {cv_results['test_roc_auc'].mean():.4f}")
print(f"Mean Malignant Recall: {cv_results['test_recall_malignant'].mean():.4f}")
```

## E. Improve Accuracy

### E.1 Algorithm Tuning
**Exercise E.1.** Tune the SVM (`C`, `gamma`, `kernel`) with `GridSearchCV` inside a scaling pipeline.

```python
# TODO: grid search over C=[0.1,1,10,100], gamma=["scale",0.001,0.01,0.1], kernel=["rbf","linear"]
...

param_grid_svm = {
    'model__C': [0.1, 1, 10, 100],
    'model__gamma': ['scale', 0.001, 0.01, 0.1],
    'model__kernel': ['rbf', 'linear']
}

pipe_svm = Pipeline([
    ('scale', StandardScaler()),
    ('model', SVC(random_state=seed))
])

grid_svm = GridSearchCV(pipe_svm, param_grid_svm, cv=kfold, scoring='accuracy', n_jobs=-1)
grid_svm.fit(X_train, y_train)

print("Best SVM Parameters:", grid_svm.best_params_)
print(f"Best SVM CV Accuracy: {grid_svm.best_score_:.4f}")
```

**Exercise E.2.** Tune KNN (`n_neighbors`, `weights`) the same way.

```python
# TODO: grid over n_neighbors=[3,5,7,9,11,15,21] and weights=["uniform","distance"]
...

param_grid_knn = {
    'model__n_neighbors': [3, 5, 7, 9, 11, 15, 21],
    'model__weights': ['uniform', 'distance']
}

pipe_knn = Pipeline([
    ('scale', StandardScaler()),
    ('model', KNeighborsClassifier())
])

grid_knn = GridSearchCV(pipe_knn, param_grid_knn, cv=kfold, scoring='accuracy', n_jobs=-1)
grid_knn.fit(X_train, y_train)

print("Best KNN Parameters:", grid_knn.best_params_)
print(f"Best KNN CV Accuracy: {grid_knn.best_score_:.4f}")
```

### E.2 Ensembles
**Exercise E.3.** Spot-check four ensemble methods (Random Forest, Extra Trees, Gradient Boosting, AdaBoost) and one voting ensemble of the tuned SVM, tuned KNN and Random Forest.

```python
# TODO: dictionary ensembles, score with spot_check(scale=False), then a hard VotingClassifier
...

ensembles = {
    'RF': RandomForestClassifier(random_state=seed),
    'ET': ExtraTreesClassifier(random_state=seed),
    'GBM': GradientBoostingClassifier(random_state=seed),
    'AB': AdaBoostClassifier(random_state=seed)
}

results_ensembles = spot_check(ensembles, X_train, scale=False)

best_svm_model = grid_svm.best_estimator_
best_knn_model = grid_knn.best_estimator_
rf_model = RandomForestClassifier(random_state=seed)

voting_clf = VotingClassifier(
    estimators=[
        ('svm', best_svm_model),
        ('knn', best_knn_model),
        ('rf', rf_model)
    ],
    voting='hard'
)

voting_scores = cross_val_score(voting_clf, X_train, y_train, cv=kfold, scoring='accuracy')
results_ensembles['Voting'] = voting_scores

print("Ensemble Cross-Validation Accuracy Means:")
for name, scores in results_ensembles.items():
    print(f"{name}: {scores.mean():.4f}")
```

**Exercise E.4.** Build a leaderboard (Series sorted by mean CV accuracy) containing both tuned models, the four ensembles and the voting ensemble.

```python
# TODO: leaderboard Series
...

leaderboard_dict = {
    'Tuned SVM': grid_svm.best_score_,
    'Tuned KNN': grid_knn.best_score_,
    'Random Forest': results_ensembles['RF'].mean(),
    'Extra Trees': results_ensembles['ET'].mean(),
    'Gradient Boosting': results_ensembles['GBM'].mean(),
    'AdaBoost': results_ensembles['AB'].mean(),
    'Voting Classifier': results_ensembles['Voting'].mean()
}

leaderboard = pd.Series(leaderboard_dict).sort_values(ascending=False)
print("--- Model Leaderboard (CV Accuracy) ---")
display(leaderboard)
```

### E.3 Finalize the Model
**Exercise E.5.** Fit the winner on the full training set and, for the first and only time, evaluate it on the held-out **test set**. Report accuracy, the classification report and a confusion matrix.

```python
# TODO: pick the leaderboard winner, fit on train, predict test, print metrics, plot confusion matrix
...

best_model = grid_svm.best_estimator_
best_model.fit(X_train, y_train)

y_pred = best_model.predict(X_test)

print("Test Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=data.target_names))

ConfusionMatrixDisplay.from_estimator(best_model, X_test, y_test, display_labels=data.target_names)
plt.title("Confusion Matrix (Test Set)")
plt.show()
```

#**ANSWERS:**


1. **LEAKAGE**

*Feature selection on the full dataset causes data leakage because information from the validation/test folds influences which features are selected, artificially inflating cross-validation performance. To prevent leakage, feature selection must be embedded directly inside a Pipeline (e.g., combining SelectKBest or RFE with the estimator). This ensures that feature selection is fitted strictly on the training fold of each cross-validation split and evaluated on unseen validation folds.*

2. **METRIC CHOICE**

I would report Recall (Sensitivity) for the malignant class (alongside ROC-AUC) to a clinician. In medical diagnostics, accuracy alone is insufficient because it treats all misclassifications equally. A false positive (flagging a benign tumor as malignant) leads to unnecessary follow-up tests, whereas a false negative (misdiagnosing cancer as benign) delays critical treatment, which can be life-threatening. Recall specifically measures the proportion of actual malignant cases correctly identified, directly minimizing false negatives.

3. **MODEL CHOICE**

If two models differ by only 0.3% in cross-validation accuracy, I would prioritize variance across folds, model complexity/interpretability, and computational cost over the marginal score difference:

Variance: Choose the model with lower standard deviation across folds for more stable, reliable generalization.

Interpretability: In healthcare, simpler or more interpretable models (e.g., Logistic Regression or SVM) are preferred over complex black-box ensembles if performance is essentially tied.

Efficiency: Select the model with lower training/inference latency and memory requirements for deployment.

4. **GENERALIZATION**

Smaller Dataset: A much smaller dataset increases the risk of overfitting, makes cross-validation scores more volatile (high variance), and reduces the reliability of feature selection. High-bias models (like Naive Bayes or Linear Discriminant Analysis) or strong regularization would be necessary.

Highly Imbalanced Dataset: Standard accuracy becomes practically meaningless (e.g., predicting 99% majority class gives 99% accuracy). Evaluation must shift entirely to PR-AUC (Precision-Recall AUC), F1-Score, or Balanced Accuracy, and techniques like SMOTE oversampling, undersampling, or adjusting class weights (class_weight='balanced') must be applied.
