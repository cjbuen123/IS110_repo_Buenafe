## **Exercise 2.1**

Load to a dataframe the file `GDP(PPP)_LifeExpectancy.xlsx` in the folder `Datasets`.
- Don't load the column Code which contains the full name of the country
- Use the following column names: 'Country', 'Life_Exp', 'GDP', 'Population', 'Inc_Class'
- Encode invalid values to `np.nan`

```python
import pandas as pd
import numpy as np

# Define the file path
file_path = '/content/drive/MyDrive/ColabNotebooks/ICT110-AppDev/ColabNotes/Datasets/GDP(PPP)_LifeExpectancy.xlsx'

# Load the Excel file:
# - We read columns 0, 2, 3, 4, 5 to skip the 'Code' column (which is at index 1)
# - Assign names: 'Country', 'Life_Exp', 'GDP', 'Population', 'Inc_Class'
# - Map invalid strings like '..' to np.nan using na_values
df_gdp = pd.read_excel(
    file_path,
    usecols=[0, 2, 3, 4, 5],
    names=['Country', 'Life_Exp', 'GDP', 'Population', 'Inc_Class'],
    na_values=['..']
)

# Display the loaded DataFrame
display(df_gdp.head())
```
